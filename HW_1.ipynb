**Q7. Sum of weights**

```python
df.head()
```

```python
df_asia = df[df['origin']=='Asia']
```

```python
df_asia = df_asia[['vehicle_weight', 'model_year']]
```

```python
df_asia = df_asia.head(7)
```

```python
X = df_asia.to_numpy()
X
```

```python
# X^T X
XTX = X.T @ X
```

```python
# Inverse of XTX
XTX_inv = np.linalg.inv(XTX)
```

```python
#  y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
# w = (X^T X)^(-1) X^T y
w = XTX_inv @ X.T @ y
```

```python
# Sum of all elements
w.sum()
```
